```python
import pandas as pd

# 1. Load the data
# Note the forward slash (/) instead of backslash (\) for the path
file_path = "C:/Users/sushm/OneDrive/Desktop/SEM 5/ML/SKYSAFE/data/processed/SkySafeDB.csv"
df = pd.read_csv(file_path)

# 2. Check shape
print("=== SHAPE ===")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

# 3. Check target balance
print("\n=== TARGET (IS_DAMAGE) BALANCE ===")
print(df['IS_DAMAGE'].value_counts(dropna=False))

# 4. Check missing values
print("\n=== MISSING VALUES ===")
print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np

# 1. Drop PRECIPITATION (too many missing values)
df = df.drop(columns=['PRECIPITATION'])

# 2. Fill categorical missing values with "Unknown"
cat_cols = ['SIZE', 'TYPE_ENG', 'PHASE_OF_FLIGHT', 'TIME_OF_DAY', 'STATE', 'SKY']
for col in cat_cols:
    df[col] = df[col].fillna('Unknown')

# 3. Fill numerical missing values with the median
num_cols = ['HEIGHT', 'SPEED', 'AC_MASS', 'NUM_ENGS']
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

# 4. For dashboard display column (Cost Repairs), fill with 0
df['COST_REPAIRS'] = df['COST_REPAIRS'].fillna(0)

# 5. Verify there are no more missing values (except maybe in COST_REPAIRS if you want)
print("Missing values after cleaning:")
print(df.isnull().sum())

# 6. Save the cleaned data using the full path
df.to_csv(r"C:\Users\sushm\OneDrive\Desktop\SEM 5\ML\SKYSAFE\data\processed\SkySafeDB_cleaned.csv", index=False)
print("\n Cleaned data saved successfully!")
```
